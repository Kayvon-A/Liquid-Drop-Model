```python
#initialize network
model = models.Sequential([
    layers.Input(shape=(6,)),         # 6 input features
    layers.Dense(32, activation='relu'),  # Hidden Layer 1 with 32 neurons
    layers.Dense(16, activation='relu'),  # Hidden Layer 2 with 16 neurons
    layers.Dense(1, activation='linear')  # Output layer for regression
])
```

```python
model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)
```
