```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_error
```

```python
w_init = np.array((15.75, -17.8, -0.711, -23.7,11.18))
```

```python
data = pd.read_csv("Randomized Data - Copy.csv")
A = data.pop("A")
Z = data.pop("Z")
N = data.pop("N")
y = np.array((data.pop("BE")))
BE_A = data.pop("BE/A")
x= np.array((data))
print(x.shape)
```

```python
LDMD = []
for i in range(x.shape[0]):
    tmp = np.dot(x[i], w_init)
    LDMD.append(tmp)
```

```python
#Initial
plt.scatter(A, LDMD, color="red")
plt.scatter(A, y, color="blue")
```

```python
LDMD_A = []
for i in range(x.shape[0]):
    tmp = np.dot(x[i], w_init) /A[i]
    LDMD_A.append(tmp)
```

```python
plt.scatter(A, LDMD_A, color="red")
plt.scatter(A, BE_A, color="blue")
```

```python
mae = mean_absolute_error(y, LDMD)
mse = mean_squared_error(y, LDMD)
print(mae)
print(mse)
```

```python
mae = mean_absolute_error(BE_A, LDMD_A)
mse = mean_squared_error(BE_A, LDMD_A)
print(mae)
print(mse)
```

```python
w_pred =  np.array((15.25561479, -16.2500258,   -0.68837504, -22.17336291,  11.48861481))
```

```python
predictions = []
for i in range(x.shape[0]):
    v = np.dot(x[i],w_pred)
    predictions.append(v)
```

```python
plt.scatter(A, predictions, color="red")
plt.scatter(A, y, color="blue")
```

```python
predictions_A = []
for i in range(x.shape[0]):
    v = np.dot(x[i],w_pred) /A[i]
    predictions_A.append(v)
```

```python
plt.scatter(A, predictions_A, color="red")
plt.scatter(A, BE_A, color="blue")
```

```python
mae = mean_absolute_error(y, predictions)
mse = mean_squared_error(y, predictions )
print(mae)
print(mse)
```

```python
mae = mean_absolute_error(BE_A, predictions_A)
mse = mean_squared_error(BE_A, predictions_A)
print(mae)
print(mse)
```
